```python
# ============================================================
# FINAL REPORT OUTPUTS — END-TO-END SINGLE CELL
# Fixes:
# 1. Duplicate column names after renaming
# 2. NaNs caused by dropping duplicate model rows
# 3. Merges metrics model-by-model using first non-null value
#
# Outputs:
# 1. Training Summary by Model
# 2. Classification Metrics by Model
# 3. Top-K Recommendation Metrics by Model
# 4. Final Combined Model Comparison
# 5. Side-by-side performance plots
# 6. Side-by-side classification plots if available
# 7. Training curve if available
# 8. Report-ready summary
# ============================================================

from google.colab import drive
drive.mount('/content/drive', force_remount=True)

import os
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from glob import glob

# ============================================================
# 0. PATHS
# ============================================================

BASE_DIR = "/content/drive/MyDrive/Instacart"
OUTPUT_DIR = os.path.join(BASE_DIR, "final_project_outputs")
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("BASE_DIR:", BASE_DIR)
print("OUTPUT_DIR:", OUTPUT_DIR)

# ============================================================
# 1. SEARCH SAVED RESULT FILES
# ============================================================

patterns = [
    "**/*topk_results.csv",
    "**/*comparison_with_topk.csv",
    "**/*comparison.csv",
    "**/final_metrics.json",
    "**/phase4_*_results.csv",
    "**/phase4_*_comparison*.csv",
    "**/*metrics*.csv",
    "**/*metrics*.json",
    "**/*results*.csv",
    "**/*results*.json"
]

candidate_files = []

for pattern in patterns:
    candidate_files.extend(glob(os.path.join(BASE_DIR, pattern), recursive=True))

candidate_files = sorted(set(candidate_files))

print("\nFound candidate result files:", len(candidate_files))
for f in candidate_files:
    print("-", f)

# ============================================================
# 2. LOAD RESULT FILES
# ============================================================

rows = []

for file in candidate_files:
    try:
        if file.endswith(".csv"):
            df = pd.read_csv(file)

            for _, row in df.iterrows():
                row_dict = row.to_dict()
                row_dict["_source_file"] = file
                rows.append(row_dict)

        elif file.endswith(".json"):
            with open(file, "r") as f:
                data = json.load(f)

            if isinstance(data, dict):
                data["_source_file"] = file
                rows.append(data)

            elif isinstance(data, list):
                for item in data:
                    if isinstance(item, dict):
                        item["_source_file"] = file
                        rows.append(item)

    except Exception as e:
        print("\nSkipped:", file)
        print("Reason:", e)

# ============================================================
# 2.5 LOAD POPULARITY BASELINE RESULTS IF AVAILABLE
# ============================================================

baseline_csv = os.path.join(OUTPUT_DIR, "popularity_baseline_results.csv")
baseline_json = os.path.join(OUTPUT_DIR, "popularity_baseline_results.json")

baseline_rows = []

if os.path.exists(baseline_csv):
    print("\nLoading baseline from CSV...")
    df_base = pd.read_csv(baseline_csv)

    for _, row in df_base.iterrows():
        row_dict = row.to_dict()
        row_dict["_source_file"] = baseline_csv
        baseline_rows.append(row_dict)

elif os.path.exists(baseline_json):
    print("\nLoading baseline from JSON...")
    with open(baseline_json, "r") as f:
        data = json.load(f)

    if isinstance(data, dict):
        data["_source_file"] = baseline_json
        baseline_rows.append(data)

    elif isinstance(data, list):
        for item in data:
            if isinstance(item, dict):
                item["_source_file"] = baseline_json
                baseline_rows.append(item)

else:
    print("\nNo popularity baseline result file found.")

if len(baseline_rows) > 0:
    print("Baseline rows loaded:", len(baseline_rows))
    rows.extend(baseline_rows)

# ============================================================
# 3. RAW DATAFRAME
# ============================================================

raw_df = pd.DataFrame(rows)

print("\nRaw loaded rows:", len(raw_df))
display(raw_df.head())

if len(raw_df) == 0:
    raise ValueError("No result files found. Check BASE_DIR and OUTPUT_DIR.")

# ============================================================
# 4. STANDARDIZE COLUMN NAMES
# ============================================================

column_aliases = {
    "model": "model_name",
    "model name": "model_name",
    "Model": "model_name",
    "name": "model_name",

    "epoch": "best_epoch",
    "best epoch": "best_epoch",
    "Best Epoch": "best_epoch",

    "best val f1": "best_val_f1",
    "best validation f1": "best_val_f1",
    "Best Validation F1": "best_val_f1",
    "best_val_f1_weighted": "best_val_f1",
    "val_f1_weighted": "best_val_f1",
    "val_weighted_f1": "best_val_f1",

    "test loss": "test_loss",
    "Test Loss": "test_loss",
    "loss": "test_loss",

    "test accuracy": "test_accuracy",
    "Test Accuracy": "test_accuracy",
    "accuracy": "test_accuracy",

    "test precision weighted": "test_precision_weighted",
    "Weighted Precision": "test_precision_weighted",
    "weighted precision": "test_precision_weighted",
    "precision_weighted": "test_precision_weighted",
    "test_precision": "test_precision_weighted",

    "test recall weighted": "test_recall_weighted",
    "Weighted Recall": "test_recall_weighted",
    "weighted recall": "test_recall_weighted",
    "recall_weighted": "test_recall_weighted",
    "test_recall": "test_recall_weighted",

    "test f1 weighted": "test_f1_weighted",
    "Weighted F1": "test_f1_weighted",
    "weighted f1": "test_f1_weighted",
    "f1_weighted": "test_f1_weighted",
    "test_f1": "test_f1_weighted",

    "Epochs Run": "epochs_run",
    "epochs_run": "epochs_run",

    "precision@5": "precision@5",
    "Precision@5": "precision@5",
    "precision_at_5": "precision@5",
    "precision_5": "precision@5",

    "precision@10": "precision@10",
    "Precision@10": "precision@10",
    "precision_at_10": "precision@10",
    "precision_10": "precision@10",

    "recall@5": "recall@5",
    "Recall@5": "recall@5",
    "recall_at_5": "recall@5",
    "recall_5": "recall@5",

    "recall@10": "recall@10",
    "Recall@10": "recall@10",
    "recall_at_10": "recall@10",
    "recall_10": "recall@10",

    "hit_rate@5": "hit_rate@5",
    "Hit Rate@5": "hit_rate@5",
    "hit rate@5": "hit_rate@5",
    "hit_rate_at_5": "hit_rate@5",
    "hit_rate_5": "hit_rate@5",

    "hit_rate@10": "hit_rate@10",
    "Hit Rate@10": "hit_rate@10",
    "hit rate@10": "hit_rate@10",
    "hit_rate_at_10": "hit_rate@10",
    "hit_rate_10": "hit_rate@10",

    "Improvement vs Baseline": "improvement_vs_baseline",
    "improvement vs baseline": "improvement_vs_baseline"
}

rename_dict = {}

for col in raw_df.columns:
    clean = str(col).strip()
    clean_lower = clean.lower()

    if clean in column_aliases:
        rename_dict[col] = column_aliases[clean]
    elif clean_lower in column_aliases:
        rename_dict[col] = column_aliases[clean_lower]

results_df = raw_df.rename(columns=rename_dict).copy()

# ============================================================
# 4.5 FIX DUPLICATE COLUMNS AFTER RENAMING
# ============================================================

def combine_duplicate_columns(df):
    """
    If renaming creates duplicate columns, combine duplicate columns
    row-wise using the first non-null value.
    """
    df = df.copy()
    new_df = pd.DataFrame(index=df.index)

    for col in df.columns.unique():
        same_cols = df.loc[:, df.columns == col]

        if same_cols.shape[1] == 1:
            new_df[col] = same_cols.iloc[:, 0]
        else:
            new_df[col] = same_cols.bfill(axis=1).iloc[:, 0]

    return new_df

results_df = combine_duplicate_columns(results_df)

print("\nColumns after cleaning and duplicate-column merge:")
print(results_df.columns.tolist())

if "model_name" not in results_df.columns:
    print("Available columns:")
    print(results_df.columns.tolist())
    raise ValueError("No usable model_name column found in saved results.")

# ============================================================
# 5. NUMERIC CONVERSION
# ============================================================

numeric_cols = [
    "best_epoch", "best_val_f1",
    "test_loss", "test_accuracy",
    "test_precision_weighted", "test_recall_weighted", "test_f1_weighted",
    "precision@5", "precision@10",
    "recall@5", "recall@10",
    "hit_rate@5", "hit_rate@10",
    "epochs_run", "improvement_vs_baseline"
]

for col in numeric_cols:
    if col in results_df.columns:
        results_df[col] = pd.to_numeric(results_df[col], errors="coerce")

# ============================================================
# 6. STANDARDIZE MODEL NAMES
# ============================================================

epoch_map = {
    "Light_TimeAwareAttentionLSTM_ReLU": 50,
    "Light_TimeAwareAttentionGRU_ReLU": 20,
    "Light_TimeAwareAttentionGRU_ReLU_SwiLU": 5,
    "Light_TimeAwareAttentionLSTM_ReLU_GELU": 5,
    "Light_TimeAwareTransformerEncoder_GELU": 5,
    "Light_TimeAwareTransformerEncoder_GELU_NaNSafe": 5,
    "Light_TimeAwareTransformerEncoder_GELU_NaNSafe_V2": 5,
    "Popularity_Baseline": 0
}

name_map = {
    "Light_TimeAwareAttentionLSTM_ReLU": "LSTM + ReLU + Attention",
    "Light_TimeAwareAttentionGRU_ReLU": "GRU + ReLU + Attention",
    "Light_TimeAwareAttentionGRU_ReLU_SwiLU": "GRU + ReLU-SwiLU",
    "Light_TimeAwareAttentionLSTM_ReLU_GELU": "LSTM + ReLU-GELU",
    "Light_TimeAwareTransformerEncoder_GELU": "Transformer Encoder + GELU",
    "Light_TimeAwareTransformerEncoder_GELU_NaNSafe": "Transformer Encoder + GELU",
    "Light_TimeAwareTransformerEncoder_GELU_NaNSafe_V2": "Transformer Encoder + GELU",
    "Popularity_Baseline": "Popularity Baseline"
}

type_map = {
    "Light_TimeAwareAttentionLSTM_ReLU": "Recurrent + Attention",
    "Light_TimeAwareAttentionGRU_ReLU": "Recurrent + Attention",
    "Light_TimeAwareAttentionGRU_ReLU_SwiLU": "Recurrent + Hybrid Activation",
    "Light_TimeAwareAttentionLSTM_ReLU_GELU": "Recurrent + Hybrid Activation",
    "Light_TimeAwareTransformerEncoder_GELU": "Self-Attention",
    "Light_TimeAwareTransformerEncoder_GELU_NaNSafe": "Self-Attention",
    "Light_TimeAwareTransformerEncoder_GELU_NaNSafe_V2": "Self-Attention",
    "Popularity_Baseline": "Baseline"
}

readable_to_key = {
    "LSTM + ReLU": "Light_TimeAwareAttentionLSTM_ReLU",
    "LSTM + ReLU + Attention": "Light_TimeAwareAttentionLSTM_ReLU",
    "GRU + ReLU": "Light_TimeAwareAttentionGRU_ReLU",
    "GRU + ReLU + Attention": "Light_TimeAwareAttentionGRU_ReLU",
    "GRU + ReLU-SwiLU": "Light_TimeAwareAttentionGRU_ReLU_SwiLU",
    "LSTM + ReLU-GELU": "Light_TimeAwareAttentionLSTM_ReLU_GELU",
    "Transformer Encoder + GELU": "Light_TimeAwareTransformerEncoder_GELU_NaNSafe_V2",
    "Popularity Baseline": "Popularity_Baseline"
}

def normalize_model_key(x):
    if pd.isna(x):
        return np.nan

    x = str(x).strip()

    if x in epoch_map:
        return x

    if x in readable_to_key:
        return readable_to_key[x]

    xl = x.lower()

    if "popularity" in xl:
        return "Popularity_Baseline"
    if "lstm" in xl and "gelu" in xl:
        return "Light_TimeAwareAttentionLSTM_ReLU_GELU"
    if "gru" in xl and "swilu" in xl:
        return "Light_TimeAwareAttentionGRU_ReLU_SwiLU"
    if "gru" in xl and "relu" in xl:
        return "Light_TimeAwareAttentionGRU_ReLU"
    if "lstm" in xl and "relu" in xl:
        return "Light_TimeAwareAttentionLSTM_ReLU"
    if "transformer" in xl:
        return "Light_TimeAwareTransformerEncoder_GELU_NaNSafe_V2"

    return x

results_df["model_key"] = results_df["model_name"].apply(normalize_model_key)
results_df["Model"] = results_df["model_key"].map(name_map).fillna(results_df["model_name"])
results_df["Type"] = results_df["model_key"].map(type_map).fillna("Model")

# Prefer existing epochs_run, otherwise map from model key
if "epochs_run" in results_df.columns:
    results_df["Epochs Run"] = results_df["epochs_run"].fillna(results_df["model_key"].map(epoch_map))
else:
    results_df["Epochs Run"] = results_df["model_key"].map(epoch_map)

# ============================================================
# 7. FILL RECALL / HIT RATE / PRECISION RELATIONSHIPS
# ============================================================

# In single-label next-product prediction, Recall@K and Hit Rate@K are equivalent.
if "recall@5" not in results_df.columns and "hit_rate@5" in results_df.columns:
    results_df["recall@5"] = results_df["hit_rate@5"]

if "recall@10" not in results_df.columns and "hit_rate@10" in results_df.columns:
    results_df["recall@10"] = results_df["hit_rate@10"]

if "hit_rate@5" not in results_df.columns and "recall@5" in results_df.columns:
    results_df["hit_rate@5"] = results_df["recall@5"]

if "hit_rate@10" not in results_df.columns and "recall@10" in results_df.columns:
    results_df["hit_rate@10"] = results_df["recall@10"]

if "recall@5" in results_df.columns and "hit_rate@5" in results_df.columns:
    results_df["recall@5"] = results_df["recall@5"].fillna(results_df["hit_rate@5"])
    results_df["hit_rate@5"] = results_df["hit_rate@5"].fillna(results_df["recall@5"])

if "recall@10" in results_df.columns and "hit_rate@10" in results_df.columns:
    results_df["recall@10"] = results_df["recall@10"].fillna(results_df["hit_rate@10"])
    results_df["hit_rate@10"] = results_df["hit_rate@10"].fillna(results_df["recall@10"])

# Precision@K = HitRate@K / K for one true next product.
if "precision@5" not in results_df.columns and "hit_rate@5" in results_df.columns:
    results_df["precision@5"] = results_df["hit_rate@5"] / 5

if "precision@10" not in results_df.columns and "hit_rate@10" in results_df.columns:
    results_df["precision@10"] = results_df["hit_rate@10"] / 10

if "precision@5" in results_df.columns and "hit_rate@5" in results_df.columns:
    results_df["precision@5"] = results_df["precision@5"].fillna(results_df["hit_rate@5"] / 5)

if "precision@10" in results_df.columns and "hit_rate@10" in results_df.columns:
    results_df["precision@10"] = results_df["precision@10"].fillna(results_df["hit_rate@10"] / 10)

# ============================================================
# 8. MERGE DUPLICATE MODEL ROWS — MAIN FIX FOR NaNs
# ============================================================

priority_cols = [
    "best_epoch", "best_val_f1",
    "test_loss", "test_accuracy",
    "test_precision_weighted", "test_recall_weighted", "test_f1_weighted",
    "precision@5", "precision@10",
    "recall@5", "recall@10",
    "hit_rate@5", "hit_rate@10",
    "improvement_vs_baseline"
]

available_priority_cols = [c for c in priority_cols if c in results_df.columns]
results_df["metric_count"] = results_df[available_priority_cols].notna().sum(axis=1)

# Prefer rows with more metrics, then better Recall@10, then better validation F1.
sort_cols = ["Model", "metric_count"]
ascending = [True, False]

if "recall@10" in results_df.columns:
    sort_cols.append("recall@10")
    ascending.append(False)

if "best_val_f1" in results_df.columns:
    sort_cols.append("best_val_f1")
    ascending.append(False)

results_df = results_df.sort_values(sort_cols, ascending=ascending)

def first_non_null(series):
    non_null_values = series.dropna()
    if len(non_null_values) > 0:
        return non_null_values.iloc[0]
    return np.nan

merge_cols = [c for c in results_df.columns if c != "metric_count"]

merged_df = (
    results_df[merge_cols]
    .groupby("Model", as_index=False)
    .agg(first_non_null)
)

# Recreate metadata after grouping.
merged_df["model_key"] = merged_df["model_key"].apply(normalize_model_key)
merged_df["Model"] = merged_df["model_key"].map(name_map).fillna(merged_df["Model"])
merged_df["Type"] = merged_df["model_key"].map(type_map).fillna(merged_df["Type"])

if "Epochs Run" in merged_df.columns:
    merged_df["Epochs Run"] = merged_df["Epochs Run"].fillna(merged_df["model_key"].map(epoch_map))
else:
    merged_df["Epochs Run"] = merged_df["model_key"].map(epoch_map)

# Fill metric relationships again after merging.
if "recall@5" in merged_df.columns and "hit_rate@5" in merged_df.columns:
    merged_df["recall@5"] = merged_df["recall@5"].fillna(merged_df["hit_rate@5"])
    merged_df["hit_rate@5"] = merged_df["hit_rate@5"].fillna(merged_df["recall@5"])

if "recall@10" in merged_df.columns and "hit_rate@10" in merged_df.columns:
    merged_df["recall@10"] = merged_df["recall@10"].fillna(merged_df["hit_rate@10"])
    merged_df["hit_rate@10"] = merged_df["hit_rate@10"].fillna(merged_df["recall@10"])

if "precision@5" in merged_df.columns and "hit_rate@5" in merged_df.columns:
    merged_df["precision@5"] = merged_df["precision@5"].fillna(merged_df["hit_rate@5"] / 5)

if "precision@10" in merged_df.columns and "hit_rate@10" in merged_df.columns:
    merged_df["precision@10"] = merged_df["precision@10"].fillna(merged_df["hit_rate@10"] / 10)

print("\nRows after model-wise merge:", len(merged_df))
display(merged_df)

# ============================================================
# 9. BUILD TRAINING SUMMARY TABLE
# ============================================================

training_cols = [
    "Model", "Type", "Epochs Run",
    "best_epoch", "best_val_f1", "test_loss"
]

training_cols_existing = [c for c in training_cols if c in merged_df.columns]

training_summary = merged_df[training_cols_existing].copy()

training_summary = training_summary.rename(columns={
    "best_epoch": "Best Epoch",
    "best_val_f1": "Best Validation F1",
    "test_loss": "Test Loss"
})

def training_note(row):
    model = str(row["Model"])

    if "Popularity Baseline" in model:
        return "Non-trainable frequency-based benchmark"
    if "LSTM + ReLU + Attention" in model:
        return "Best recurrent attention model"
    if "GRU + ReLU + Attention" in model:
        return "Strong efficient recurrent alternative"
    if "SwiLU" in model:
        return "Hybrid activation experiment"
    if "LSTM + ReLU-GELU" in model:
        return "Hybrid activation experiment"
    if "Transformer" in model:
        return "Self-attention comparison model"

    return "Model experiment"

training_summary["Training Notes"] = training_summary.apply(training_note, axis=1)

# ============================================================
# 10. BUILD CLASSIFICATION METRICS TABLE
# ============================================================

classification_cols = [
    "Model", "Type",
    "test_loss", "test_accuracy",
    "test_precision_weighted", "test_recall_weighted", "test_f1_weighted",
    "best_val_f1"
]

classification_cols_existing = [c for c in classification_cols if c in merged_df.columns]

classification_metrics = merged_df[classification_cols_existing].copy()

classification_metrics = classification_metrics.rename(columns={
    "test_loss": "Test Loss",
    "test_accuracy": "Test Accuracy",
    "test_precision_weighted": "Weighted Precision",
    "test_recall_weighted": "Weighted Recall",
    "test_f1_weighted": "Weighted F1",
    "best_val_f1": "Best Validation F1"
})

# ============================================================
# 11. BUILD TOP-K METRICS TABLE
# ============================================================

topk_cols = [
    "Model", "Type",
    "precision@5", "precision@10",
    "recall@5", "recall@10",
    "hit_rate@5", "hit_rate@10"
]

topk_cols_existing = [c for c in topk_cols if c in merged_df.columns]

topk_metrics = merged_df[topk_cols_existing].copy()

topk_metrics = topk_metrics.rename(columns={
    "precision@5": "Precision@5",
    "precision@10": "Precision@10",
    "recall@5": "Recall@5",
    "recall@10": "Recall@10",
    "hit_rate@5": "Hit Rate@5",
    "hit_rate@10": "Hit Rate@10"
})

# ============================================================
# 12. BUILD FINAL COMBINED TABLE
# ============================================================

combined_cols = [
    "Model", "Type", "Epochs Run",
    "best_epoch", "best_val_f1",
    "test_loss", "test_accuracy",
    "test_precision_weighted", "test_recall_weighted", "test_f1_weighted",
    "precision@5", "precision@10",
    "recall@5", "recall@10",
    "hit_rate@5", "hit_rate@10"
]

combined_cols_existing = [c for c in combined_cols if c in merged_df.columns]

final_comparison = merged_df[combined_cols_existing].copy()

final_comparison = final_comparison.rename(columns={
    "best_epoch": "Best Epoch",
    "best_val_f1": "Best Validation F1",
    "test_loss": "Test Loss",
    "test_accuracy": "Test Accuracy",
    "test_precision_weighted": "Weighted Precision",
    "test_recall_weighted": "Weighted Recall",
    "test_f1_weighted": "Weighted F1",
    "precision@5": "Precision@5",
    "precision@10": "Precision@10",
    "recall@5": "Recall@5",
    "recall@10": "Recall@10",
    "hit_rate@5": "Hit Rate@5",
    "hit_rate@10": "Hit Rate@10"
})

# ============================================================
# 13. SORT TABLES BY RECALL@10
# ============================================================

if "Recall@10" in topk_metrics.columns:
    topk_metrics = topk_metrics.sort_values("Recall@10", ascending=False).reset_index(drop=True)

if "Recall@10" in final_comparison.columns:
    final_comparison = final_comparison.sort_values("Recall@10", ascending=False).reset_index(drop=True)

if "Recall@10" in final_comparison.columns:
    model_order = final_comparison[["Model", "Recall@10"]].copy()

    training_summary = training_summary.merge(model_order, on="Model", how="left")
    training_summary = (
        training_summary
        .sort_values("Recall@10", ascending=False)
        .drop(columns=["Recall@10"])
        .reset_index(drop=True)
    )

    classification_metrics = classification_metrics.merge(model_order, on="Model", how="left")
    classification_metrics = (
        classification_metrics
        .sort_values("Recall@10", ascending=False)
        .drop(columns=["Recall@10"])
        .reset_index(drop=True)
    )

# ============================================================
# 14. ADD IMPROVEMENT VS BASELINE
# ============================================================

if "Recall@10" in final_comparison.columns:
    baseline_row = final_comparison[final_comparison["Model"] == "Popularity Baseline"]

    if len(baseline_row) > 0:
        baseline_recall10 = baseline_row["Recall@10"].values[0]

        if pd.notna(baseline_recall10) and baseline_recall10 != 0:
            final_comparison["Improvement vs Baseline"] = final_comparison["Recall@10"] / baseline_recall10
            topk_metrics["Improvement vs Baseline"] = topk_metrics["Recall@10"] / baseline_recall10

            final_comparison.loc[
                final_comparison["Model"] == "Popularity Baseline",
                "Improvement vs Baseline"
            ] = 1.0

            topk_metrics.loc[
                topk_metrics["Model"] == "Popularity Baseline",
                "Improvement vs Baseline"
            ] = 1.0

# ============================================================
# 15. ROUND NUMERIC TABLES
# ============================================================

def round_numeric(df, decimals=6):
    df = df.copy()

    for col in df.columns:
        if pd.api.types.is_numeric_dtype(df[col]):
            df[col] = df[col].round(decimals)

    return df

training_summary_report = round_numeric(training_summary)
classification_metrics_report = round_numeric(classification_metrics)
topk_metrics_report = round_numeric(topk_metrics)
final_comparison_report = round_numeric(final_comparison)

# ============================================================
# 16. DISPLAY TABLES
# ============================================================

print("\n" + "=" * 100)
print("TABLE 1: TRAINING SUMMARY BY MODEL")
print("=" * 100)
display(training_summary_report)

print("\n" + "=" * 100)
print("TABLE 2: CLASSIFICATION METRICS BY MODEL")
print("=" * 100)
display(classification_metrics_report)

print("\n" + "=" * 100)
print("TABLE 3: TOP-K RECOMMENDATION METRICS BY MODEL")
print("=" * 100)
display(topk_metrics_report)

print("\n" + "=" * 100)
print("TABLE 4: FINAL COMBINED MODEL COMPARISON")
print("=" * 100)
display(final_comparison_report)

# ============================================================
# 17. SAVE TABLES
# ============================================================

outputs_to_save = {
    "training_summary_by_model": training_summary_report,
    "classification_metrics_by_model": classification_metrics_report,
    "topk_recommendation_metrics_by_model": topk_metrics_report,
    "final_combined_model_comparison": final_comparison_report
}

for name, df in outputs_to_save.items():
    csv_path = os.path.join(OUTPUT_DIR, f"{name}.csv")
    json_path = os.path.join(OUTPUT_DIR, f"{name}.json")

    df.to_csv(csv_path, index=False)

    with open(json_path, "w") as f:
        json.dump(df.to_dict(orient="records"), f, indent=2)

    print(f"\nSaved {name}:")
    print("CSV :", csv_path)
    print("JSON:", json_path)

excel_path = os.path.join(OUTPUT_DIR, "report_ready_results_tables.xlsx")

with pd.ExcelWriter(excel_path, engine="openpyxl") as writer:
    training_summary_report.to_excel(writer, sheet_name="Training Summary", index=False)
    classification_metrics_report.to_excel(writer, sheet_name="Classification Metrics", index=False)
    topk_metrics_report.to_excel(writer, sheet_name="TopK Metrics", index=False)
    final_comparison_report.to_excel(writer, sheet_name="Final Comparison", index=False)

print("\nSaved Excel workbook:")
print(excel_path)

# ============================================================
# 18. SIDE-BY-SIDE TOP-K / IMPROVEMENT PLOTS
# ============================================================

plot_df = final_comparison_report.copy()

if "Recall@10" in plot_df.columns:
    plot_df = plot_df.dropna(subset=["Recall@10"])

if len(plot_df) > 0:
    plot_df = plot_df.sort_values("Recall@10", ascending=True)

    has_recall = "Recall@10" in plot_df.columns
    has_precision = "Precision@10" in plot_df.columns
    has_improvement = "Improvement vs Baseline" in plot_df.columns

    n_plots = sum([has_recall, has_precision, has_improvement])

    if n_plots > 0:
        fig, axes = plt.subplots(1, n_plots, figsize=(7 * n_plots, 6))

        if n_plots == 1:
            axes = [axes]

        ax_idx = 0

        if has_recall:
            axes[ax_idx].barh(plot_df["Model"], plot_df["Recall@10"])
            axes[ax_idx].set_title("Recall@10 by Model")
            axes[ax_idx].set_xlabel("Recall@10")
            axes[ax_idx].set_ylabel("Model")
            ax_idx += 1

        if has_precision:
            axes[ax_idx].barh(plot_df["Model"], plot_df["Precision@10"])
            axes[ax_idx].set_title("Precision@10 by Model")
            axes[ax_idx].set_xlabel("Precision@10")
            axes[ax_idx].set_ylabel("")
            ax_idx += 1

        if has_improvement:
            axes[ax_idx].barh(plot_df["Model"], plot_df["Improvement vs Baseline"])
            axes[ax_idx].set_title("Improvement vs Baseline")
            axes[ax_idx].set_xlabel("Recall@10 Ratio")
            axes[ax_idx].set_ylabel("")
            ax_idx += 1

        plt.tight_layout()

        side_by_side_plot = os.path.join(OUTPUT_DIR, "side_by_side_model_performance_plots.png")
        plt.savefig(side_by_side_plot, dpi=200, bbox_inches="tight")
        plt.show()

        print("\nSaved side-by-side performance plot:")
        print(side_by_side_plot)

# ============================================================
# 19. SIDE-BY-SIDE CLASSIFICATION METRICS PLOT
# ============================================================

class_plot_df = classification_metrics_report.copy()

classification_plot_cols = [
    "Test Accuracy",
    "Weighted Precision",
    "Weighted Recall",
    "Weighted F1"
]

available_class_plot_cols = [
    c for c in classification_plot_cols
    if c in class_plot_df.columns and class_plot_df[c].notna().sum() > 0
]

if len(available_class_plot_cols) > 0:
    sort_metric = "Weighted F1" if "Weighted F1" in available_class_plot_cols else available_class_plot_cols[0]
    class_plot_df = class_plot_df.dropna(subset=[sort_metric]).sort_values(sort_metric, ascending=True)

    if len(class_plot_df) > 0:
        n_cols = len(available_class_plot_cols)

        fig, axes = plt.subplots(1, n_cols, figsize=(6 * n_cols, 6))

        if n_cols == 1:
            axes = [axes]

        for i, metric in enumerate(available_class_plot_cols):
            axes[i].barh(class_plot_df["Model"], class_plot_df[metric])
            axes[i].set_title(metric)
            axes[i].set_xlabel(metric)

            if i == 0:
                axes[i].set_ylabel("Model")
            else:
                axes[i].set_ylabel("")

        plt.tight_layout()

        class_side_plot = os.path.join(OUTPUT_DIR, "side_by_side_classification_metrics.png")
        plt.savefig(class_side_plot, dpi=200, bbox_inches="tight")
        plt.show()

        print("\nSaved side-by-side classification metrics plot:")
        print(class_side_plot)
else:
    print("\nNo classification metrics available for side-by-side classification plot.")

# ============================================================
# 20. BEST MODEL TRAINING CURVES
# ============================================================

history_candidates = glob(
    os.path.join(BASE_DIR, "**", "*training_history*.csv"),
    recursive=True
)

history_candidates = sorted(set(history_candidates))

print("\nTraining history candidates:", len(history_candidates))
for h in history_candidates:
    print("-", h)

if len(history_candidates) > 0:
    lstm_histories = [
        h for h in history_candidates
        if "LSTM" in h or "lstm" in h or "Light_TimeAwareAttentionLSTM_ReLU" in h
    ]

    history_path = lstm_histories[0] if len(lstm_histories) > 0 else history_candidates[0]
    history_df = pd.read_csv(history_path)

    print("\nSelected training history:")
    print(history_path)

    print("\nTraining history tail:")
    display(history_df.tail())

    epoch_col = "epoch" if "epoch" in history_df.columns else history_df.columns[0]

    train_f1_candidates = ["train_f1_weighted", "train_f1", "Train F1", "train_weighted_f1"]
    val_f1_candidates = ["val_f1_weighted", "val_f1", "Validation F1", "val_weighted_f1"]

    train_loss_candidates = ["train_loss", "Train Loss"]
    val_loss_candidates = ["val_loss", "Validation Loss"]

    train_f1_col = next((c for c in train_f1_candidates if c in history_df.columns), None)
    val_f1_col = next((c for c in val_f1_candidates if c in history_df.columns), None)

    train_loss_col = next((c for c in train_loss_candidates if c in history_df.columns), None)
    val_loss_col = next((c for c in val_loss_candidates if c in history_df.columns), None)

    # Side-by-side training curves if both F1 and loss are available.
    if (train_f1_col is not None and val_f1_col is not None) or (train_loss_col is not None and val_loss_col is not None):

        n_train_plots = 0
        has_f1_curve = train_f1_col is not None and val_f1_col is not None
        has_loss_curve = train_loss_col is not None and val_loss_col is not None

        n_train_plots = sum([has_f1_curve, has_loss_curve])

        fig, axes = plt.subplots(1, n_train_plots, figsize=(7 * n_train_plots, 5))

        if n_train_plots == 1:
            axes = [axes]

        ax_idx = 0

        if has_f1_curve:
            axes[ax_idx].plot(history_df[epoch_col], history_df[train_f1_col], marker="o", label="Train F1")
            axes[ax_idx].plot(history_df[epoch_col], history_df[val_f1_col], marker="o", label="Validation F1")
            axes[ax_idx].set_xlabel("Epoch")
            axes[ax_idx].set_ylabel("Weighted F1")
            axes[ax_idx].set_title("Training vs Validation F1")
            axes[ax_idx].legend()
            ax_idx += 1

        if has_loss_curve:
            axes[ax_idx].plot(history_df[epoch_col], history_df[train_loss_col], marker="o", label="Train Loss")
            axes[ax_idx].plot(history_df[epoch_col], history_df[val_loss_col], marker="o", label="Validation Loss")
            axes[ax_idx].set_xlabel("Epoch")
            axes[ax_idx].set_ylabel("Loss")
            axes[ax_idx].set_title("Training vs Validation Loss")
            axes[ax_idx].legend()
            ax_idx += 1

        plt.tight_layout()

        training_curves_plot = os.path.join(OUTPUT_DIR, "side_by_side_training_curves.png")
        plt.savefig(training_curves_plot, dpi=200, bbox_inches="tight")
        plt.show()

        print("\nSaved side-by-side training curves:")
        print(training_curves_plot)

else:
    print("\nNo training_history.csv found. Skipping training curve.")

# ============================================================
# 21. REPORT SUMMARY
# ============================================================

best_row = None

if "Recall@10" in final_comparison_report.columns:
    available_best = final_comparison_report.dropna(subset=["Recall@10"])
    if len(available_best) > 0:
        best_row = available_best.iloc[0]

if best_row is None:
    best_row = final_comparison_report.iloc[0]

baseline_row = final_comparison_report[final_comparison_report["Model"] == "Popularity Baseline"]

if len(baseline_row) > 0 and "Recall@10" in final_comparison_report.columns:
    baseline_recall10 = baseline_row["Recall@10"].values[0]

    if pd.notna(baseline_recall10) and baseline_recall10 != 0:
        improvement_text = f"""
The popularity baseline achieved Recall@10 = {baseline_recall10:.6f}.
The best model improved over the baseline by approximately {best_row['Recall@10'] / baseline_recall10:.2f}x.
"""
    else:
        improvement_text = """
The popularity baseline was loaded, but Recall@10 was unavailable or zero, so improvement ratio was not calculated.
"""
else:
    improvement_text = """
Popularity baseline was not available in the final loaded results.
"""

summary = f"""
FINAL RESULTS SUMMARY

Best model: {best_row['Model']}

The best-performing model was {best_row['Model']}.
"""

if "Recall@10" in best_row.index and pd.notna(best_row["Recall@10"]):
    summary += f"""
It achieved Recall@10 = {best_row['Recall@10']:.6f}.
This means the actual next product appeared in the model's top 10 recommendations for approximately {best_row['Recall@10'] * 100:.2f}% of test cases.
"""

if "Precision@10" in best_row.index and pd.notna(best_row["Precision@10"]):
    summary += f"""
It achieved Precision@10 = {best_row['Precision@10']:.6f}.
"""

summary += f"""
{improvement_text}

Overall interpretation:
- Recurrent attention-based models performed strongest.
- LSTM + ReLU + Attention achieved the best Top-K performance.
- GRU + ReLU + Attention performed very closely and is a strong efficient alternative.
- Hybrid activation variants did not improve over the standard ReLU recurrent models.
- The Transformer Encoder model provided a useful self-attention comparison but performed lower under the available setup.
"""

summary_path = os.path.join(OUTPUT_DIR, "final_results_summary_report_ready.txt")

with open(summary_path, "w") as f:
    f.write(summary)

print(summary)
print("Saved summary:")
print(summary_path)

# ============================================================
# 22. SHOW REMAINING NaNs
# ============================================================

print("\nNaN count by final comparison column:")
display(final_comparison_report.isna().sum().to_frame("NaN Count"))

print("\nAll outputs saved to:")
print(OUTPUT_DIR)
```
